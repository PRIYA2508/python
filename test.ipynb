```python
#3. String Comparison Operators -> compare two strings
#(i) Comparison are case sensitive ("A" != "a").ASCII value compared to each other.ASCII value of capital letters is less than small letters
#.  (1) == ,(2) != , (3)< ,(4)<= ,(5)> ,(6)>= ,(7)
print("Hello"=="hello")
print("hello" == "hEllo") #here capatial E doesn't equal to small e
print("hello" != "Hello")
print("hello"<"HELLO")
print("HELLO"<"hello")
print("hello"<= "hello")
```

```python
#4.Membership operator
#(i)In operator -> gives true when first operand contain inside second operand
print('PRI' in 'priya')
print('pri' in 'priya')

#(ii)not in operator -> gives true when first operand not present in second operand basically vice-versa of in operator
print('hju' not in 'priya')
print('pri' not in 'priya')
```

```python
#6. inetpolation -> interopolation is the intersetion of something of a different nature into something else.
#string interpolation -> The process of inserting an object into predefined string is called string interpolation or string formatting.

#string formatting operator (%)-> it helps us to including external object in our string or (placeholder).
# %d = decimal formatter , %c = include external charcater , %s= external string in string , %f= floating point in string
address = "shanti nagar gate no-2"
name = "priya"
print("this is my address %s" % (address))
print("my name is %s and my address is %s" %(name,address))
```
